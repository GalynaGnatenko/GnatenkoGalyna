3. Цікаво, який тип клієнтів приносить найбільший прибуток. Теж розподілення майже однакове. Але все ж таки члену клубу магазину приносять прибутку більше. Тож компанії потрібно переводити пересічних клієнтів на членів клубу магазину.

```python
pt = pd.pivot_table(df,     index='Customer type',     values='gross income',     aggfunc='sum',     margins=True,     margins_name='Total')
pt['% of total'] = (pt['gross income'] / pt.loc['Total', 'gross income'] * 100).round(2)

pt
```
