# Тести для малих вибірок та пропорцій



**Завдання 1**. E-commerce компанія після редизайну сайту підозрює, що **середній час до покупки (time-to-purchase)** користувачів **збільшився**.
Історично середній час від першого заходу на сайт до покупки становив $\mu_0 = 123.7$ хвилин.

Команда аналітиків випадково обрала дані **7 покупців після редизайну**:
`128, 135, 121, 142, 126, 151, 123`

З допомогою наявних даних зʼясуйте, чи збільшився середній час покупки після редизайну.

Для цього виконайте наступні 6 кроків. Правильне виконання кожного з кроків оцінюється в 1 бал.

1. Запишіть нульову та альтернативну гіпотези і визначте тип тесту.
2. Обчисліть вибіркові статистики: $\bar x$, $s$, $n$.
3. Оберіть тип тесту та виконайте його (знайдіть критичне значення тесту, статистику тесту та р-значення) будь-яким пасуючим способом, щоб перевірити гіпотезу на рівні значущості $\alpha = 0.10$.
4. Прийміть рішення, чи відхиляєте ви гіпотезу $H_0$ використовуючи p-value.
5. Напишіть висновок: чи справді редизайн сайту подовжив час до покупки?
6. Чи зміниться ваше рішення при зміні рівня значущості на $\alpha = 0.05$.


```python
import scipy.stats as stats
import numpy as np

# Дані
data = [128, 135, 121, 142, 126, 151, 123]
mu_0 = 123.7
alpha_1 = 0.10
alpha_2 = 0.05
```

```python
# 1. Запишіть нульову та альтернативну гіпотези і визначте тип тесту
print(f'Завдання 1')
print(f"Нульова гіпотеза H0: mu_0  = {mu_0}")
print(f"Альтернативна гіпотеза Ha: mu > {mu_0}")
print("Тип тесту: Правосторонній тест.")
```

```python
#2. Обчисліть вибіркові статистики:  x¯ ,  s ,  n .
print(f'Завдання 2')
# Крок 2: Статистики
n = len(data)
x_bar = np.mean(data)
s = np.std(data, ddof=1)
print(f"Обсяг вибірки: {n}")
print(f"Середнє (x_bar): {x_bar:.2f}")
print(f"Стандартне відхилення (s): {s:.2f}")
```

```python
#3 Оберіть тип тесту та виконайте його (знайдіть критичне значення тесту, статистику тесту та р-значення) будь-яким пасуючим способом, щоб перевірити гіпотезу на рівні значущості  α=0.10 .
print(f"Завдання 3")
alpha = 0.10
df = n - 1
# t-статистика та p-value за допомогою scipy
t_stat, p_val_two_tailed = stats.ttest_1samp(data, mu_0)

# Оскільки тест правосторонній (Ha: mu > mu_0), ділимо двостороннє p-value навпіл:
p_value = p_val_two_tailed / 2 if t_stat > 0 else 1 - (p_val_two_tailed / 2)

# Критичне значення для правостороннього t-тесту
t_crit = stats.t.ppf(1 - alpha, df)

print(f"Рівень значущості (alpha): {alpha}")
print(f"Ступені свободи (df): {df}")
print(f"t-статистика: {t_stat:.2f}")
print(f"Критичне значення t_crit: {t_crit:.2f}")
print(f"p-значення (p-value): {p_value:.4f}")
```

```python
# 4, 5, 6. Прийняття рішення, висновок та перевірка для alpha = 0.05

alpha_1 = 0.10
alpha_2 = 0.05

# Пункт 4. Прийняття рішення при alpha = 0.10
print(f"Завдання 4")
print(f"Оскільки p-value ({p_value:.4f}) < alpha ({alpha_1}), ми ВІДХИЛЯЄМО нульову гіпотезу H0. покупці стали витрачати більше часу на покупку. Це не випадковість і не збіг обставин — зміни на сайті реально на це вплинули")
print("-" * 50)

# Пункт 5. Висновок
print(f"Завдання 5")
print("Висновок: На рівні значущості alpha = 0.10 є статистично значущі докази того, що редизайн сайту справді подовжив середній час до покупки.")
print("-" * 50)

# Пункт 6. Перевірка при alpha = 0.05
print(f"Завдання 6")
print(f"При alpha = {alpha_2}: p-value ({p_value:.4f}) < {alpha_2}. Рішення НЕ ЗМІНИТЬСЯ — ми все одно відхиляємо H0.")
```

**Завдання 2.**

До спеціальної рекламної кампанії **23%** дорослих упізнавали логотип компанії. Після завершення кампанії відділ маркетингу провів опитування: з **1200** випадково відібраних дорослих **311** упізнали логотип.

Перевірте на рівні значущості **3%** ($\alpha=0.03$), чи дають ці дані достатні підстави стверджувати, що **тепер більше ніж 23%** дорослих упізнають логотип компанії. Для розвʼязку використовуйте бібліотеку `statsmodels`.

Зробіть висновок, чи зросла впізнаваність логотипу.

Додатково, обчисліть довірчий інтревал на заданому рівні значущості і проінтерпретуйте текстом - як він додатково пояснює прийняте нами рішення?

```python
import statsmodels.api as sm
import numpy as np

n = 1200        # загальна кількість опитаних дорослих
count = 311     # кількість тих, хто впізнав логотип
p_0 = 0.23      # початкова частка (23%)
alpha = 0.03    # рівень значущості (3%)

# 1. Одновибірковий Z-тест для частки (правосторонній: alternative='larger')
z_stat, p_value = sm.stats.proportions_ztest(count, n, value=p_0, alternative='larger')

# 2. Обчислення довірчого інтервалу (рівень довіри 1 - alpha = 0.97 або 97%)
ci_low, ci_upp = sm.stats.proportion_confint(count, n, alpha=alpha, method='normal')

# Виведення результатів у вашому стилі
print("=== Завдання 2. Перевірка гіпотези для пропорцій ===")
print(f"Z-статистика: {z_stat:.4f}")
print(f"p-value: {p_value:.4f}")
print(f"Рівень значущості (alpha): {alpha}\n")

# Прийняття рішення та висновок
if p_value < alpha:
    print(f"Рішення: Оскільки p-value ({p_value:.4f}) < alpha ({alpha}), ми ВІДХИЛЯЄМО нульову гіпотезу H0.")
    print("Висновок: Так, дані свідчать про те, що впізнаваність логотипу статистично значуще зросла і перевищує 23%.")
else:
    print(f"Рішення: Оскільки p-value ({p_value:.4f}) >= alpha ({alpha}), ми НЕ ВІДХИЛЯЄМО нульову гіпотезу H0.")
    print("Висновок: Недостатньо доказів стверджувати, що впізнаваність зросла.")

print("-" * 50)
print(f"97% Довірчий інтервал для частки: [{ci_low:.4f}, {ci_upp:.4f}]  ({ci_low*100:.2f}% — {ci_upp*100:.2f}%)")
```
